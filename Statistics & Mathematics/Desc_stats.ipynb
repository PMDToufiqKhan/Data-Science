## Descriptive Statistics

```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
# Sample dataset
data = [15, 18, 22, 20, 18, 25, 20, 19, 21, 18, 100]  # 100 is outlier
```

#### Using Numpy

```python
# Mean
mean = np.mean(data)
print(f"Mean: {mean:.2f}")
```

```python
# Median
median = np.median(data)
print(f"Median: {median:.2f}")
```

```python
# Mode
from scipy.stats import mode
mode_result = stats.mode(data, keepdims=True)
print(f"Mode: {mode_result.mode[0]}")
```

```python
# Range
range_val = max(data) - min(data)
print(f"Range: {range_val}")
```

```python
# Variance (population)                     
variance_pop = np.var(data)                  # by default ddof = 0 which means squared units are divided by N
print(f"Variance (Population): {variance_pop:.2f}")
```

```python
# Variance (sample)
variance_sample = np.var(data, ddof=1)       # Delta Degrees of Freedom (ddof) = 1 means divide by N-1
print(f"Variance (Sample): {variance_sample:.2f}")
```

```python
# Standard Deviation
std_dev = np.std(data, ddof=1)
print(f"Standard Deviation: {std_dev:.2f}")
```

```python
# Coefficient of Variation
cv = (std_dev / mean) * 100
print(f"Coefficient of Variation: {cv:.2f}%")
```

#### Using Pandas

```python
df = pd.DataFrame(data)
print(df.describe())
```

## QUARTILES & IQR

```python
# Using numpy
print(np.percentile(data, 25))
print(np.percentile(data, 50))
print(np.percentile(data, 75))
```

```python
# Using pandas
print(df.quantile(0.25))
print(df.quantile(0.50))
print(df.quantile(0.75))
```

## OUTLIER DETECTION

```python
# IQR method using numpy. same thing can also be done using pandas (quantile method)
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in data if x < lower_bound or x > upper_bound]
for outlier in outliers:
    print(outlier)
```

## Weighted Averages

```python
# Example 1: GPA Calculation
print("\nExample 1: Student GPA")
grades = [4.0, 3.0, 3.5]
credits = [4, 3, 3]

gpa = np.average(grades, weights=credits)
print(f"Grades: {grades}")
print(f"Credit Hours: {credits}")
print(f"Weighted GPA: {gpa:.2f}")

# Example 2: Portfolio Weighted Price
print("\nExample 2: Portfolio Weighted Price")
prices = [20, 30, 40]
quantities = [100, 50, 25]

weighted_price = np.average(prices, weights=quantities)
print(f"Prices: {prices}")
print(f"Quantities: {quantities}")
print(f"Weighted Average Price: ${weighted_price:.2f}")
```

## Cumulative sum

```python
# Example 1: Daily Sales Cumulative
print("\nExample 1: Daily Sales Running Total")
daily_sales = [1000, 1500, 2000, 1200, 800]

cumsum = np.cumsum(daily_sales)

df = pd.DataFrame({
    'Day': range(1, len(daily_sales) + 1),
    'Daily Sales': daily_sales,
    'Cumulative Sales': cumsum
})

print(df)
print(f"\nTotal Sales: ${cumsum[-1]}")
```
